<h2>Data Cleaning</h2>

<b>'Teacher_Quality', 'Parental_Education_Level' and 'Distance_from_Home' have null values</b>

```python
df_student['Teacher_Quality'].value_counts()
```

```python
df_student['Teacher_Quality'] = df_student['Teacher_Quality'].fillna('Medium')
df_student.isna().sum()
```

```python
df_student['Teacher_Quality'].value_counts()
```

```python
df_student['Parental_Education_Level'].value_counts()
```

```python
df_student['Parental_Education_Level'] = df_student['Parental_Education_Level'].fillna('High School')
df_student.isna().sum()
```

```python
df_student['Distance_from_Home'].value_counts()
```

```python
df_student['Distance_from_Home'] = df_student['Distance_from_Home'].fillna('Near')
df_student['Distance_from_Home'].value_counts()
```

<b>No more null values</b>

```python
df_student.isna().sum()
```

<h2>Exploratory Data Analysis (EDA) & Visulization</h2>

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style('darkgrid')

# Distribution of Exam Scores
plt.figure(figsize=(10, 6))
sns.histplot(df_student['Exam_Score'], kde=True, bins= 20)
plt.title('Distribution of Exam Scores')
plt.xlabel('Exam Score')
plt.ylabel('Frequency')
plt.show()
```

<h4>Relationship between Study Hours and Exam Score</h4>

```python
plt.figure(figsize=(10, 6))
sns.regplot(df_student, x='Hours_Studied', y='Exam_Score', )
plt.title('Exam Score vs. Hours Studied')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.show()
```

```python
df_student['Parental_Education_Level'].value_counts()
```

<h4>Impact of 'Parental_Education_Level' on Exam Score</h4>

```python
plt.figure(figsize=(12, 7))
sns.boxplot(df_student, x='Parental_Education_Level', y='Exam_Score', order=['High School', 'College', 'Postgraduate'])
plt.title('Exam Score by Parental Education Level', fontweight='bold')
plt.xlabel('Parental Education Level')
plt.ylabel('Exam Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

<h4>Correlation Heatmap (for numerical features)</h4>

```python
numerical_cols = df_student.select_dtypes(include=['int64', 'float64']).columns
correlation_matrix = df_student[numerical_cols].corr()
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features', fontweight='bold')
plt.show()
```
